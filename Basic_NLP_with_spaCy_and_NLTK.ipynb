```python
# Import required libraries
import spacy
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tag import pos_tag
from collections import Counter
import pandas as pd

# Download NLTK data (do this once)
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('wordnet')
nltk.download('stopwords')
nltk.download('maxent_ne_chunker')
nltk.download('words')

# Load spaCy model
nlp = spacy.load('en_core_web_sm')

print("✅ All libraries loaded successfully!")
```

## Part 2: Tokenization

**What is Tokenization?**

Breaking down text into smaller chunks (tokens) - usually words or sentences.

**Analogy**: If a paragraph is a cake, tokenization is cutting it into slices (sentences) and then cutting each slice into crumbs (words).

```python
# Sample text
sample_text = "Dr. Smith lives in New York. He loves playing cricket and coding!"

print("Original text:")
print(sample_text)
print("\n" + "="*60 + "\n")

# 1. NLTK - Word Tokenization
print("NLTK - Word Tokenization:")
words_nltk = word_tokenize(sample_text)
print(words_nltk)
print(f"Number of tokens: {len(words_nltk)}")

print("\n" + "="*60 + "\n")

# 2. spaCy - Word Tokenization
print("spaCy - Word Tokenization:")
doc = nlp(sample_text)
words_spacy = [token.text for token in doc]
print(words_spacy)
print(f"Number of tokens: {len(words_spacy)}")
```

```python
# 3. NLTK - Sentence Tokenization
print("NLTK - Sentence Tokenization:")
sentences_nltk = sent_tokenize(sample_text)
for i, sent in enumerate(sentences_nltk, 1):
    print(f"Sentence {i}: {sent}")

print("\n" + "="*60 + "\n")

# 4. spaCy - Sentence Tokenization
print("spaCy - Sentence Tokenization:")
sentences_spacy = [sent.text for sent in doc.sents]
for i, sent in enumerate(sentences_spacy, 1):
    print(f"Sentence {i}: {sent}")
```

**Key Difference**: spaCy handles punctuation and abbreviations (like "Dr.") better than NLTK.

---
## Part 3: Removing Stopwords

**What are Stopwords?**

Common words (the, is, and, a, etc.) that don't add much meaning to text. Removing them helps us focus on important words.

**Analogy**: Like removing water from tea to get the actual tea leaves - the leaves (meaningful words) are what matter!

```python
# Get English stopwords
stop_words = set(stopwords.words('english'))

print(f"Total stopwords: {len(stop_words)}")
print(f"\nFirst 20 stopwords: {list(stop_words)[:20]}")

print("\n" + "="*60 + "\n")

# Remove stopwords using NLTK
print("Before removing stopwords:")
print(words_nltk)

filtered_words = [word for word in words_nltk if word.lower() not in stop_words]
print(f"\nAfter removing stopwords:")
print(filtered_words)

print(f"\nRemoved {len(words_nltk) - len(filtered_words)} stopwords!")
```

```python
# Using spaCy to remove stopwords
print("Using spaCy to identify stopwords:")
print("\nToken\t\t| Is Stopword?")
print("-" * 40)
for token in doc:
    print(f"{token.text:15} | {token.is_stop}")
```

---
## Part 4: Stemming and Lemmatization

**What's the difference?**

- **Stemming**: Removes prefixes/suffixes to get the root word. (e.g., "running" → "runn")
- **Lemmatization**: Converts words to their dictionary form. (e.g., "running" → "run")

**Analogy**: Stemming is like trimming a plant with scissors (crude), while lemmatization is like properly pruning it with knowledge (smart).

```python
# Sample words to process
words_to_process = ["running", "runs", "ran", "loved", "loving", "loves", "better", "better"]

print("STEMMING vs LEMMATIZATION")
print("=" * 60)

# Initialize tools
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

print(f"\n{'Original':12} | {'Stemmed':12} | {'Lemmatized':12}")
print("-" * 40)

for word in words_to_process:
    stemmed = stemmer.stem(word)
    lemmatized = lemmatizer.lemmatize(word, pos='v')  # pos='v' means verb
    print(f"{word:12} | {stemmed:12} | {lemmatized:12}")
```

```python
# Using spaCy for lemmatization
text = "The cats are running faster than the dogs"
doc = nlp(text)

print("spaCy Lemmatization:")
print(f"\n{'Token':12} | {'Lemma':12} | {'POS':10}")
print("-" * 40)
for token in doc:
    print(f"{token.text:12} | {token.lemma_:12} | {token.pos_:10}")
```

---
## Part 5: Part-of-Speech (POS) Tagging

**What is POS Tagging?**

Labeling each word with its grammatical role (noun, verb, adjective, etc.).

**Example**: In "The cat runs", "cat" is a NOUN, "runs" is a VERB.

```python
# Using NLTK for POS tagging
print("NLTK - Part-of-Speech Tagging:")
print("=" * 60)

text = "The quick brown fox jumps over the lazy dog"
tokens = word_tokenize(text)
pos_tags = pos_tag(tokens)

print(f"\n{'Word':10} | {'POS Tag':10}")
print("-" * 25)
for word, tag in pos_tags:
    print(f"{word:10} | {tag:10}")
```

```python
# Using spaCy for POS tagging
print("spaCy - Part-of-Speech Tagging:")
print("=" * 60)

doc = nlp(text)

print(f"\n{'Word':10} | {'POS':8} | {'Explanation':30}")
print("-" * 55)
for token in doc:
    pos_name = spacy.explain(token.pos_)  # Get explanation
    print(f"{token.text:10} | {token.pos_:8} | {pos_name:30}")
```

**POS Tag Meanings:**
- **NOUN**: Nouns (cat, dog, house)
- **VERB**: Verbs (run, jump, eat)
- **ADJ**: Adjectives (quick, lazy, big)
- **DET**: Determiners (the, a, an)
- **ADP**: Adpositions (in, on, over)

## Part 6: Named Entity Recognition (NER)

**What is NER?**

Identifying and classifying named entities (people, organizations, locations, dates, etc.) in text.

**Example**: In "Elon Musk works at Tesla in California", we identify:
- **Elon Musk** → PERSON
- **Tesla** → ORG
- **California** → GPE (location)

```python
# Using spaCy for Named Entity Recognition
print("spaCy - Named Entity Recognition (NER)")
print("=" * 70)

text = """Apple Inc. was founded by Steve Jobs in Cupertino, California. 
The company released the iPhone on June 29, 2007."""

doc = nlp(text)

print(f"\nText: {text}")
print("\n" + "-" * 70)
print(f"\n{'Entity':20} | {'Label':10} | {'Description':30}")
print("-" * 70)

for ent in doc.ents:
    description = spacy.explain(ent.label_)
    print(f"{ent.text:20} | {ent.label_:10} | {description:30}")
```

**Common Entity Labels:**
- **PERSON**: People (Steve Jobs, Elon Musk)
- **ORG**: Organizations (Apple, Google, Tesla)
- **GPE**: Geographic/Political entities (USA, California, Paris)
- **DATE**: Dates (June 29, 2007, tomorrow)
- **PRODUCT**: Products (iPhone, Windows)
- **MONEY**: Money/Currency amounts ($100, €50)

## Part 7: Text Cleaning and Preprocessing Pipeline

Let's combine everything into a practical preprocessing pipeline!

```python
def preprocess_text(text):
    """
    Complete text preprocessing pipeline
    
    Steps:
    1. Tokenize
    2. Remove stopwords
    3. Lemmatize
    4. Remove punctuation
    5. Convert to lowercase
    """
    
    # Process with spaCy
    doc = nlp(text.lower())  # Convert to lowercase
    
    # Process each token
    processed_tokens = []
    for token in doc:
        # Skip stopwords and punctuation
        if token.is_stop or token.is_punct:
            continue
        # Keep lemmatized form
        processed_tokens.append(token.lemma_)
    
    return processed_tokens


# Test the pipeline
original_text = """The machine learning algorithms are really powerful for analyzing data! 
                    They help us process information faster and better."""

print("ORIGINAL TEXT:")
print(original_text)

print("\n" + "=" * 70)
print("\nAFTER PREPROCESSING:")

processed = preprocess_text(original_text)
print(processed)

print("\n" + "-" * 70)
print(f"Original tokens: {len(word_tokenize(original_text))}")
print(f"Processed tokens: {len(processed)}")
print(f"Removed: {len(word_tokenize(original_text)) - len(processed)} tokens")
```

---
## Part 8: Word Frequency Analysis

Analyzing which words appear most frequently in a text.

```python
# Sample document
document = """Cricket is a popular sport. Cricket players need practice. 
              The cricket team trains daily. Cricket is fun!"""

print("Document:")
print(document)

print("\n" + "=" * 70)

# Preprocess
processed = preprocess_text(document)
print("\nProcessed words:")
print(processed)

# Count frequency
word_freq = Counter(processed)

print("\n" + "-" * 70)
print("\nWORD FREQUENCY (Top 10):")
print("\n" + f"{'Word':15} | {'Frequency':10}")
print("-" * 30)

for word, count in word_freq.most_common(10):
    print(f"{word:15} | {count:10}")
```

```python
# Visualize with pandas
import matplotlib.pyplot as plt

# Create DataFrame
freq_df = pd.DataFrame(word_freq.most_common(10), columns=['Word', 'Frequency'])

# Display as table
print(freq_df.to_string(index=False))

# Plot
plt.figure(figsize=(10, 6))
plt.bar(freq_df['Word'], freq_df['Frequency'], color='steelblue')
plt.xlabel('Words')
plt.ylabel('Frequency')
plt.title('Word Frequency Distribution')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Part 9: Complete Example - Analyzing a News Article

Let's put everything together!

```python
# Sample news article
article = """Apple Inc. announced new products on Monday. CEO Tim Cook presented the latest iPhone 
             and Apple Watch in California. The company expects strong sales in the coming months. 
             Investors are excited about the announcements made during the keynote event."""

print("📰 ARTICLE ANALYSIS")
print("=" * 80)
print(f"\nOriginal Article:\n{article}")

print("\n" + "=" * 80)
print("\n1️⃣  SENTENCE SEGMENTATION")
print("-" * 80)

doc = nlp(article)
for i, sent in enumerate(doc.sents, 1):
    print(f"Sentence {i}: {sent.text}")

print("\n" + "=" * 80)
print("\n2️⃣  NAMED ENTITIES")
print("-" * 80)

for ent in doc.ents:
    print(f"🏷️  {ent.text:20} → {ent.label_:10} ({spacy.explain(ent.label_)})")

print("\n" + "=" * 80)
print("\n3️⃣  KEY WORDS (After Preprocessing)")
print("-" * 80)

processed = preprocess_text(article)
key_words = Counter(processed).most_common(8)

for word, count in key_words:
    bar = "█" * count
    print(f"{word:15} {bar} ({count})")

print("\n" + "=" * 80)
print("\n4️⃣  POS TAGGING (First 15 tokens)")
print("-" * 80)

print(f"\n{'Word':15} | {'POS':8} | {'Description':35}")
print("-" * 65)

for token in list(doc)[:15]:
    pos_name = spacy.explain(token.pos_)
    print(f"{token.text:15} | {token.pos_:8} | {pos_name:35}")
```

## 🎓 Key Takeaways

1. **Tokenization**: Breaking text into words/sentences

4. **POS Tagging**: Understanding word types (noun, verb, etc.)
   - Useful for grammar analysis

5. **NER**: Finding and classifying named entities
   - Extract important information like people, places, organizations
